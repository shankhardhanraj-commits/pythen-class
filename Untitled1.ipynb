```python
a=np.array([1,4,9,16])
print("square root:",np.sqrt(a))
print("log:",np.log(a))
print("exp:",np.exp(a))
print("sin:",np.sin(a))
print("cos:",np.cos(a))
```

```python
data= np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])
print("mean of all values:")
print(data.mean())

print("\nmean of each colum:")
print(data.mean(axis=0))

print("\nmean of each row:")
print(data.mean(axis=1))
```
