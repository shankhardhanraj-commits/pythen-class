```python
destination=input("where is your destination?:")
participants=int(input("Number of participants:"))
    
transportation = 12000
admission_per_person=800
meal_per_person =1000
available_budget =35000
if participants <= 0:
    print("Invalid number of participants.")
else:
    admission_total = admission_per_person * participants
    meals_total = meal_per_person * participants
                              
    total = transportation + admission_total + meals_total
    budget_remaining = available_budget - total
    within_budget = total <= available_budget                       
    print("Destination:", destination)
    print("Participants:", participants, "people")
    print("Transportation:", transportation, "yen")
    print("Admission total:", admission_total, "yen")
    print("Meals total:", meals_total, "yen")
    print("Total trip cost:", total, "yen")
    print("Available budget:", available_budget, "yen")
    print("Budget remaining:", budget_remaining, "yen")
    print("Within budget:", within_budget)
```
