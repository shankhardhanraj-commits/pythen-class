```python
# 1. Even Number Accumulator

limit = int(input("Positive integer: "))

while limit <= 0:
    print("Please enter a positive integer.")
    limit = int(input("Positive integer: "))

number = 2
total = 0

while number <= limit:
    print(number)
    total += number
    number += 2

print("Total:", total)
```

```python
# 2. Limited PIN Attempts

secret = "2468"
attempts = 0
access_granted = False

while attempts < 3:
    pin = input("Enter PIN: ")
    attempts += 1

    if pin == secret:
        access_granted = True
        print("Access granted in", attempts, "attempts")
        break

    print("Incorrect PIN")

if not access_granted:
    print("Access denied")
```

```python
# 3. Skip Negative Inputs

total = 0
count = 0

while True:
    number = int(input("Number (0 to stop): "))

    if number == 0:
        break

    if number < 0:
        continue

    total += number
    count += 1

print("Count:", count, "Total:", total)
```

```python
# 4. Interactive Score Menu

count = 0
total = 0

while True:
    print("1 Add  2 Summary  0 Exit")
    choice = input("Choice: ")

    if choice == "0":
        print("Goodbye!")
        break

    elif choice == "1":
        score = int(input("Score: "))

        while score < 0 or score > 100:
            print("Invalid score. Enter a value from 0 to 100.")
            score = int(input("Score: "))

        total += score
        count += 1

    elif choice == "2":
        if count == 0:
            print("No scores yet")
        else:
            print("Scores:", count, "Average:", total / count)

    else:
        print("Invalid choice")
        continue
```
