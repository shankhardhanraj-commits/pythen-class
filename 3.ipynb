```python
students=[
    {
        "name":"dhanraj",
        "grade":"A",
        "marks":90
    },
    {"name":"ram","grade":"B", "marks":95},
    {"name":"hari","grade":"C", "marks":80},
    {"name":"sita","grade":"D", "marks":40}
 ]
    
```

```python
for student in students:
    print(student["name"],student["grade"],student["marks"])
```

```python
for student in students:
    if student["marks"]>=60:
        print(student["name"], ":","pass")
    else:
        print(student["name"],"fail")
```

```python
marks=[45,65,78,55,78]
passed=0
fail=0
for mark in marks:
    if mark>=60:
        passed+=1
    else:
        fail +=1
print("passed",passed)
print("failed",fail)
```
