```python
name=input("enter your name")
marks=int(input("enter your marks"))
if marks >=90:
    grade="A"
elif marks >=80:
    grade="B"
elif marks >=70:
    grade="C"
elif marks >=60:
    grade="D"
else:
    grade="fail"

print("name:",name)
print("marks:",marks)
print("grade:",grade)
```
