```python
students=[
    {"name":"ram","marks":90},
    {"name":"dhanraj","marks":99},
    {"name":"ananta","marks":80},
    {"name":"sita","marks":68},
    {"name":"gita","marks":45}
]
marks=[student["marks"] for student in students]
print("Total student:",len(students))
print("heights mark:",max(marks))
print("lowest mark:",min(marks))
print("avrage mark:",sum(marks)/len(marks))
print("result:")

for student in students:
    if student["marks"] >=60:
        print(student["name"], "pass")
    else:
        print(student["name"], "fail")

```

```python
students=[
    {"name":"ram","marks":90},
    {"name":"dhanraj","marks":99},
    {"name":"ananta","marks":80},
    {"name":"sita","marks":68},
    {"name":"gita","marks":45}
]

passed=0
failed=0
for student in students:
    if student["marks"] >60:
        passed+=1
    else:
        failed +=1
topstudent=max(students, key=lambda s:s["marks"])
passparcentage=passed/len(students)*100
        
print("passed",passed)
print("failed",failed)
print("pass parcentage:",passparcentage)
print("top student:", topstudent["name"])
```
